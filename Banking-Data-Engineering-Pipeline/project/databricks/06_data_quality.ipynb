```python
# ============================================================
# BANKING DATA ENGINEERING PROJECT
# FINAL GOLD DATA QUALITY
# ============================================================

from pyspark.sql import SparkSession
from pyspark.sql.functions import col


# ============================================================
# SPARK SESSION
# ============================================================

spark = SparkSession.builder.appName(
    "Banking_Final_Data_Quality"
).getOrCreate()


# ============================================================
# GOLD PATHS
# ============================================================

gold_account_path = (
    "/Volumes/bank_data/default/bank_data/"
    "gold_account_summary"
)

gold_daily_path = (
    "/Volumes/bank_data/default/bank_data/"
    "gold_daily_banking_kpis"
)

gold_transaction_path = (
    "/Volumes/bank_data/default/bank_data/"
    "gold_transaction_analysis"
)


# ============================================================
# EXPECTED VALUES
# ============================================================

EXPECTED_ACCOUNT_ROWS = 10
EXPECTED_ACCOUNT_COLUMNS = 7

EXPECTED_DAILY_ROWS = 1294
EXPECTED_DAILY_COLUMNS = 7

EXPECTED_TRANSACTION_ROWS = 2
EXPECTED_TRANSACTION_COLUMNS = 4


# ============================================================
# LOAD GOLD DATA
# ============================================================

print("========================================")
print("        LOAD GOLD DATA")
print("========================================")


account_df = (
    spark.read
    .format("delta")
    .load(gold_account_path)
)

daily_df = (
    spark.read
    .format("delta")
    .load(gold_daily_path)
)

transaction_df = (
    spark.read
    .format("delta")
    .load(gold_transaction_path)
)


print("All Gold datasets loaded successfully")


# ============================================================
# TABLE EXISTENCE / LOAD CHECK
# ============================================================

print("\n========================================")
print("        GOLD DATASET CHECK")
print("========================================")


print(" gold_account_summary exists")
print(" gold_daily_banking_kpis exists")
print(" gold_transaction_analysis exists")


# ============================================================
# ROW COUNT CHECK
# ============================================================

print("\n========================================")
print("          ROW COUNT CHECK")
print("========================================")


account_rows = account_df.count()
daily_rows = daily_df.count()
transaction_rows = transaction_df.count()


print(
    f"Account Summary: {account_rows}"
)

print(
    f"Daily KPIs: {daily_rows}"
)

print(
    f"Transaction Analysis: {transaction_rows}"
)


if account_rows == EXPECTED_ACCOUNT_ROWS:

    print(
        " Account Summary row count PASSED"
    )

else:

    raise Exception(
        f" Account Summary expected "
        f"{EXPECTED_ACCOUNT_ROWS}, got {account_rows}"
    )


if daily_rows == EXPECTED_DAILY_ROWS:

    print(
        " Daily KPIs row count PASSED"
    )

else:

    raise Exception(
        f" Daily KPIs expected "
        f"{EXPECTED_DAILY_ROWS}, got {daily_rows}"
    )


if transaction_rows == EXPECTED_TRANSACTION_ROWS:

    print(
        "Transaction Analysis row count PASSED"
    )

else:

    raise Exception(
        f" Transaction Analysis expected "
        f"{EXPECTED_TRANSACTION_ROWS}, got {transaction_rows}"
    )


# ============================================================
# COLUMN COUNT CHECK
# ============================================================

print("\n========================================")
print("         COLUMN COUNT CHECK")
print("========================================")


account_columns = len(account_df.columns)
daily_columns = len(daily_df.columns)
transaction_columns = len(transaction_df.columns)


print(
    f"Account Summary Columns: {account_columns}"
)

print(
    f"Daily KPIs Columns: {daily_columns}"
)

print(
    f"Transaction Analysis Columns: {transaction_columns}"
)


if account_columns == EXPECTED_ACCOUNT_COLUMNS:

    print(
        " Account Summary columns PASSED"
    )

else:

    raise Exception(
        " Account Summary column count FAILED"
    )


if daily_columns == EXPECTED_DAILY_COLUMNS:

    print(
        " Daily KPIs columns PASSED"
    )

else:

    raise Exception(
        " Daily KPIs column count FAILED"
    )


if transaction_columns == EXPECTED_TRANSACTION_COLUMNS:

    print(
        "Transaction Analysis columns PASSED"
    )

else:

    raise Exception(
        "Transaction Analysis column count FAILED"
    )


# ============================================================
# ACCOUNT SUMMARY NULL CHECK
# ============================================================

print("\n========================================")
print("      ACCOUNT SUMMARY NULL CHECK")
print("========================================")


null_account_no = (
    account_df
    .filter(
        col("Account_No").isNull()
    )
    .count()
)


null_account_transactions = (
    account_df
    .filter(
        col("Transaction_Count").isNull()
    )
    .count()
)


print(
    f"NULL Account_No: {null_account_no}"
)

print(
    f"NULL Transaction_Count: "
    f"{null_account_transactions}"
)


if null_account_no == 0:

    print(
        "Account Summary NULL check PASSED"
    )

else:

    raise Exception(
        "Account Summary Account_No NULL check FAILED"
    )


if null_account_transactions == 0:

    print(
        "Account Summary Transaction_Count NULL check PASSED"
    )

else:

    raise Exception(
        "Account Summary Transaction_Count NULL check FAILED"
    )


# ============================================================
# DAILY KPI NULL CHECK
# ============================================================

print("\n========================================")
print("        DAILY KPIs NULL CHECK")
print("========================================")


null_transaction_date = (
    daily_df
    .filter(
        col("Transaction_Date").isNull()
    )
    .count()
)


null_daily_transactions = (
    daily_df
    .filter(
        col("Transaction_Count").isNull()
    )
    .count()
)


print(
    f"NULL Transaction_Date: "
    f"{null_transaction_date}"
)

print(
    f"NULL Transaction_Count: "
    f"{null_daily_transactions}"
)


if null_transaction_date == 0:

    print(
        " Daily KPIs Transaction_Date NULL check PASSED"
    )

else:

    raise Exception(
        " Daily KPIs Transaction_Date NULL check FAILED"
    )


if null_daily_transactions == 0:

    print(
        " Daily KPIs Transaction_Count NULL check PASSED"
    )

else:

    raise Exception(
        " Daily KPIs Transaction_Count NULL check FAILED"
    )


# ============================================================
# TRANSACTION ANALYSIS NULL CHECK
# ============================================================

print("\n========================================")
print("     TRANSACTION ANALYSIS NULL CHECK")
print("========================================")


null_transaction_type = (
    transaction_df
    .filter(
        col("Transaction_Type").isNull()
    )
    .count()
)


null_transaction_count = (
    transaction_df
    .filter(
        col("Transaction_Count").isNull()
    )
    .count()
)


null_total_amount = (
    transaction_df
    .filter(
        col("Total_Amount").isNull()
    )
    .count()
)


print(
    f"NULL Transaction_Type: "
    f"{null_transaction_type}"
)

print(
    f"NULL Transaction_Count: "
    f"{null_transaction_count}"
)

print(
    f"NULL Total_Amount: "
    f"{null_total_amount}"
)


if null_transaction_type == 0:

    print(
        "Transaction_Type NULL check PASSED"
    )

else:

    raise Exception(
        " Transaction_Type NULL check FAILED"
    )


if null_transaction_count == 0:

    print(
        " Transaction_Count NULL check PASSED"
    )

else:

    raise Exception(
        " Transaction_Count NULL check FAILED"
    )


if null_total_amount == 0:

    print(
        " Total_Amount NULL check PASSED"
    )

else:

    raise Exception(
        " Total_Amount NULL check FAILED"
    )


# ============================================================
# DUPLICATE ACCOUNT CHECK
# ============================================================

print("\n========================================")
print("       ACCOUNT DUPLICATE CHECK")
print("========================================")


duplicate_accounts = (
    account_df
    .groupBy("Account_No")
    .count()
    .filter(
        col("count") > 1
    )
    .count()
)


print(
    f"Duplicate Account_No values: "
    f"{duplicate_accounts}"
)


if duplicate_accounts == 0:

    print(
        " Account duplicate check PASSED"
    )

else:

    raise Exception(
        " Account duplicate check FAILED"
    )


# ============================================================
# DUPLICATE DAILY DATE CHECK
# ============================================================

print("\n========================================")
print("       DAILY DATE DUPLICATE CHECK")
print("========================================")


duplicate_dates = (
    daily_df
    .groupBy("Transaction_Date")
    .count()
    .filter(
        col("count") > 1
    )
    .count()
)


print(
    f"Duplicate Transaction_Date values: "
    f"{duplicate_dates}"
)


if duplicate_dates == 0:

    print(
        " Daily date duplicate check PASSED"
    )

else:

    raise Exception(
        " Daily date duplicate check FAILED"
    )


# ============================================================
# TRANSACTION TYPE CHECK
# ============================================================

print("\n========================================")
print("      TRANSACTION TYPE CHECK")
print("========================================")


transaction_types = sorted(
    [
        row["Transaction_Type"]
        for row in (
            transaction_df
            .select("Transaction_Type")
            .distinct()
            .collect()
        )
    ]
)


print(
    f"Transaction Types: {transaction_types}"
)


expected_types = [
    "Deposit",
    "Withdrawal"
]


if transaction_types == expected_types:

    print(
        "Transaction Types check PASSED"
    )

else:

    raise Exception(
        " Transaction Types check FAILED"
    )


# ============================================================
# BUSINESS RULE CHECKS
# ============================================================

print("\n========================================")
print("        BUSINESS RULE CHECKS")
print("========================================")


invalid_account_counts = (
    account_df
    .filter(
        col("Transaction_Count") <= 0
    )
    .count()
)


invalid_daily_counts = (
    daily_df
    .filter(
        col("Transaction_Count") <= 0
    )
    .count()
)


invalid_transaction_counts = (
    transaction_df
    .filter(
        col("Transaction_Count") <= 0
    )
    .count()
)


invalid_transaction_amounts = (
    transaction_df
    .filter(
        col("Total_Amount") <= 0
    )
    .count()
)


print(
    f"Invalid Account Transaction Counts: "
    f"{invalid_account_counts}"
)

print(
    f"Invalid Daily Transaction Counts: "
    f"{invalid_daily_counts}"
)

print(
    f"Invalid Transaction Analysis Counts: "
    f"{invalid_transaction_counts}"
)

print(
    f"Invalid Transaction Amounts: "
    f"{invalid_transaction_amounts}"
)


if invalid_account_counts == 0:

    print(
        " Account business rule PASSED"
    )

else:

    raise Exception(
        " Account business rule FAILED"
    )


if invalid_daily_counts == 0:

    print(
        " Daily KPI business rule PASSED"
    )

else:

    raise Exception(
        " Daily KPI business rule FAILED"
    )


if invalid_transaction_counts == 0:

    print(
        "Transaction count business rule PASSED"
    )

else:

    raise Exception(
        " Transaction count business rule FAILED"
    )


if invalid_transaction_amounts == 0:

    print(
        " Transaction amount business rule PASSED"
    )

else:

    raise Exception(
        " Transaction amount business rule FAILED"
    )


# ============================================================
# FINAL GOLD DATA QUALITY RESULT
# ============================================================

print("\n")
print("========================================")
print("       FINAL GOLD DATA QUALITY")
print("========================================")


print(" gold_account_summary checks passed")
print(" gold_daily_banking_kpis checks passed")
print(" gold_transaction_analysis checks passed")

print("")
print("========================================")
print("   ALL GOLD DATA QUALITY CHECKS PASSED")
print("========================================")


# ============================================================
# FINAL SUMMARY
# ============================================================

print("\n========================================")
print("          PROJECT SUMMARY")
print("========================================")


print(
    f"Account Summary: "
    f"{account_rows} rows / "
    f"{account_columns} columns"
)

print(
    f"Daily Banking KPIs: "
    f"{daily_rows} rows / "
    f"{daily_columns} columns"
)

print(
    f"Transaction Analysis: "
    f"{transaction_rows} rows / "
    f"{transaction_columns} columns"
)


print("\n🎯 FINAL GOLD DATA QUALITY: PASSED")
```
